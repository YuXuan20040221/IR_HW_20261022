# 測試

先讀一篇試試


```python
import json

with open("wiki_article_list_2023_tra.json", encoding="utf-8") as f:
    for line in f:
        line = line.strip().rstrip(",")
        if line in ("[", "]"):
            continue
        article = json.loads(line)
        print(type(article))
        print(article.keys() if isinstance(article, dict) else article)
        break
```

切割字串


```python
import jieba

words = list(jieba.cut(article))
print(words[:30])
print("詞數：", len(words))
```

bigram


```python
from collections import Counter

bigrams = Counter(zip(words, words[1:]))

print("相異 bigram 數：", len(bigrams))
print("最常見的 20 個：")
print(bigrams.most_common(20))
```

讀5篇


```python
articles = []

with open("wiki_article_list_2023_tra.json", encoding="utf-8") as f:
    for line in f:
        line = line.strip().rstrip(",")
        if line in ("[", "]"):
            continue

        article = json.loads(line)
        articles.append(article)

        if len(articles) == 5:
            break

print("讀到幾篇：", len(articles))
```

```python
bigrams = Counter()
word_counts = Counter()

for article in articles:
    words = list(jieba.cut(article))
    word_counts.update(words)
    bigrams.update(Counter(zip(words, words[1:])))

print("詞彙量：", len(word_counts))
print("相異 bigrams 數：", len(bigrams))
```

# 讀取檔案

一行一行讀取json檔案的文字並斷詞。


```python
bigrams = Counter()
word_counts = Counter()
articles_num = 0

with open("wiki_article_list_2023_tra.json", encoding="utf-8") as f:
    for line in f:
        line = line.strip().rstrip(",")
        if line in ("[", "]"):
            continue

        article = json.loads(line)
        words = list(jieba.cut(article))
        word_counts.update(words)
        bigrams.update(Counter(zip(words, words[1:])))
        articles_num += 1

print("-------------------------------------------------")
print("文章數：", articles_num)
print("總詞數：", sum(word_counts.values()))
print("詞彙量：", len(word_counts))
print("相異 bigrams 數：", len(bigrams))
```

讀pkl檔


```python
import pickle

with open("wiki_counts.pkl", "rb") as f:
    saved = pickle.load(f)

word_counts = saved["word_counts"]
bigrams = saved["bigrams"]
articles_num = saved["articles_num"]
```

算機率


```python
dictionary = {}

for (first, second), count in bigrams.items():
    probability = count / word_counts[first]
    dictionary.setdefault(first, []).append((second, probability))

for matches in dictionary.values():
    matches.sort(key=lambda item: item[1], reverse=True)

with open("prob.pkl", "wb") as f:
    pickle.dump(
        {
            "dictionary": dictionary,
        },
        f,
    )
```

## 評分 prob(sentence)

回傳整句的 log 機率：把每一組相鄰詞的 log P 加起來。


```python
import math
import jieba


def prob(sentence):
    words = list(jieba.cut(sentence))
    total = 0

    for first, second in zip(words, words[1:]):
        p = 0
        for next_word, probability in dictionary.get(first, []):
            if next_word == second:
                p = probability
                break

        if p == 0:
            return float("-inf")

        total += math.log(p)

    return total


input_sentence = input("請輸入一個句子：")
print("出現【" + input_sentence + "】的 log 機率：", prob(input_sentence))
```

## 生成 generate(seed, n)

從 seed 接龍 n 個詞。取機率最高、依機率抽樣，兩種都要。

### 貪婪


```python
def generate_greedy(seed, n):
    current = seed
    result = [seed]

    for _ in range(n):
        candidates = dictionary.get(current, [])

        if not candidates:
            break

        # 取出機率最高的下一個詞
        next_word = candidates[0][0]

        result.append(next_word)
        current = next_word

    return "".join(result)


input_word = input("請輸入一個詞：")
n = int(input("請輸入要生成的詞數："))
print("生成的句子：", generate_greedy(input_word, n))
```

### 抽樣


```python
import random


def generate_gample(seed, n):
    current = seed
    result = [seed]

    for _ in range(n):
        candidates = dictionary.get(current, [])

        if not candidates:
            break

        # 隨機抽取下一個詞
        pick = random.choices(candidates, weights=[p for word, p in candidates], k=1)[0]
        next_word = pick[0]

        result.append(next_word)
        current = next_word

    return "".join(result)


input_word = input("請輸入一個詞：")
n = int(input("請輸入要生成的詞數："))
print("生成的句子：", generate_gample(input_word, n))
```
